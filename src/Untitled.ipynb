```python
import requests
import logging
from dotenv import load_dotenv
from datetime import datetime
import json
import os
from pathlib import Path
from datetime import date
```

```python
city = "Cairo"
data_folder = "data"
raw_data_folder = "raw"
```

```python
load_dotenv()

API_KEY = os.environ.get("API_KEY")

if not API_KEY:
    logging.warning("The API is missing. please make sure to enter it correctly")
    print("something went wrong_api")

url_1 = f"http://api.weatherstack.com/current?access_key={API_KEY}&query={city}"

User_Agent = os.environ.get("My_User_Agent")
if not User_Agent:
    logging.warning("The User_Agent is missing. please make sure to enter it correctly")
    print("something went wrong_User_Agent")

try:
    report = requests.get(url = url_1, headers = {"User-Agent":User_Agent}, timeout = 60)
    report.raise_for_status()
    print("rqeust is done")
except requests.exceptions as e:
    logging.warning(f"Something went wrong with the connection to the website: {e}")
```

```python
print(Path(__file__))
```

```python
data = report.json()
if not data:
    logging.warning("The data is empty")
# extracting the data into a file:
date_today_extracted = date.today()

Parent_path = Path(__file__).resolve().parent.parent/data/raw_data_folder

if Path(Parent_path).exists() is False:
    logging.warning(f"The path is not there. please create this path: {Parent_path}")

datetime_extracted = f"{city}_weather_{date_today_extracted}_extracted.json"
extracted_path = Path(Parent_path)/datetime_extracted

with open(extracted_path,"w",encoding="utf-8") as input_file:
    input_file.write(json.dumps(data))
return extracted_path
```
